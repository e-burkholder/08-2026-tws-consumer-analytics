## Objectives
The objective of this notebook is to understand the 2026 comeback *No Tragedy*, the branding/inspiration behind the album, and its success as a comeback EP.

Furthermore, this notebook aims to determine the current impressions of TWS in the domestic market while simultaneously exploring potential opportunities in the US market.

In this notebook we will:
- Research the most recent branding of TWS
- Understand the latest comeback EP
- Use the previously gathered data to draw conclusions
- Understand the current US market
- Determine potential US target audience(s)
- Analyze domestic and international trends and divergences

### Album Releases
Their music establishes a young, "boyhood pop" feeling to tell their members' stories. Since their 2024 debut, they've released 4 other albums with the most recent release being *No Tragedy* in April 2026. [(Pledis Entertainment, Discography)](https://www.pledis.co.kr/ko/artist/detail/TWS/discography/)
Album releases include:
1. Sparkling Blue (2024)
2. Summer Beat! (2024)
3. Try With Us (2025)
4. Play Hard (2025)
5. No Tragedy (2026)

Although this project focuses on analyzing the most recent album, recognizing the other albums can help create clarity and benchmarks for data analysis.

### No Tragedy
*No Tragedy* was inspired by "Romeo and Juliet" and preservering even despite the outcome. It hosts 6 tracks with a range of genres which establish a variety of music they can make as well as the flexibility their theming allows. [(ClashMusic, TWS Discuss ‘NO TRAGEDY’: A Track-By-Track Conversation)](https://www.clashmusic.com/features/tws-discuss-no-tragedy-a-track-by-track-conversation/)
