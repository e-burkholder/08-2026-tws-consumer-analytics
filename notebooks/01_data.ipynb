### Hanteo Chart
Hanteo Chart is a South Korean charting platform that tracks and measures consumer purchases of physical K-pop albums. They utilize certified retail stores and online shops to track direct business-to-consumer (B2C) sales. They often process sales transactions in real-time and show first-week initial sales records, which is what was utilized in this project.

### Hanteo
#### Exploring & Research
To start, I looked into popular sales data. I wanted to compare the album sales with their previous albums and other competitor's sales. Since Hanteo Chart tracks official and certified stores, it is widely known to reputably offer this information alongside real-time sales tracking in South Korea. I started by visiting the site, getting familiar with the layout, and searching for TWS's album sales.

After researching, I decided to utilize the cumulative album sales amounts over the initial Chodong Record, otherwise known as the sales amount over the initial 7 days release period. This data could be utilized to understand the sales trend throughout the initial launch week. In order to analyze the data, I decided to organize the information by album name, each initial day sales (1-7), and the cumulative sales number. I collected this information for *No Tragedy* alongside the two previous TWS albums *play hard* and *SUMMER BEAT!*. Since TWS doesn't operate in it's own market bubble, I also collected the information from similar competitors in the market such as BoyNextDoor (KOZ/HYBE), NCTWish (SM Entertainment), and RIIZE (SM Entertainment).

#### Collecting Data
In order to collect the data, I utilized Hanteo Chart's API to gather the public information on the website. Within the [hanteo.py](../data/py_files/hanteo.py) file, I wrote a to send a GET request to obtain the chart data associated with a specific album, indicated by the album ID, which is unique to each album on the website.

The album ids were collected by visiting the Hanteo Chart sites for each album and listing the last 9-digit number in the url. After looking at the network requests using the "Inspect" elements, I determined the request that provided the statistics for each album and copied the request as a CURL. I then utilized the public website [curlconverter](https://curlconverter.com/) to change the curl command into a Python request to implement in my .py file.

#### Headers, Parameters, Cookies
After reading through the translated request, I cleaned the headers, cookies, and parameters of any sensitive/personalized data. I put together the "idx" parameter was linked to the album_ID, which was also present in the last section of the "referer" in the header section. This was important to note as it would be adjusted for each individual album. These sections were also the only thing that changed when trying to gather information about each individual album.

#### Requesting Data & Try/Catch
The final line of the curl command was the request line from the API. I put the request into a try/except block to control possible errors and exceptions when running the command. I put in a 30 second timeout to ensure the command would stop after an extended time with no results. Assuming the response returned as normal (code 200) I imported the information into a json file, which was the raw list of data for the single album, named to reflect that.

#### Filtering Relevant Data
Once collected in json, I had to filter out the irrelevant data. Each request returned both the data for the currently viewed album, but also the information for the previous album from the artist for comparison. Since I was collecting each album individually, I determined it was best to filter out the previous album information to prevent redundant data and duplications.

To do this, I used a get request to collect the album ID, album name, daily sales, and cumulative sales from the relevant album, which was often referred to as the "Target" in the API. After extraction, the data needed to be filtered to fit a dataframe for proper analysis.

#### Organizing Data
The relevant information was now collected, but not formatted in a way that would benefit my analysis. For example, the sales data was collected as a list of numbers rather than separated by day. I created a loop that would take each entry in the list and separate it, labeling it as Day X Sales. This information was added back as its own separate column, leading to 7 days of sales data. The cumulative sales data similarly, summing the sales per day and resulting in a list of cumulative sales. Since I only needed the final cumulative(total) sales information, I separated the final value of the list as its own column.

#### Data to CSV
After the data was filtered and organized as a json I wanted to put it in a more manageable format for the project, which was determined to be a CSV (storing tabular data). I took the initial dataframe and utilized the pandas library to turn it into a table and outputting it as a CSV file.
