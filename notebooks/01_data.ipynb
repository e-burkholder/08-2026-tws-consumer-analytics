## Research Questions
When considering the objectives of this project, it is vital we collect information on a number of topics. We must determine the target consumers, as well as collect information to help determine the success of the most recent comeback campaign. Since TWS is a pre-established group, we can look at previous data insights to determine the initial target market and the actual target audience.

We can then use publicly released statistics to determine empirical values which can be used to evaluate campaign effectiveness.

Questions to ask:
- Who are the initial consumers?
- What are the demographic characterists of TWS's current audience?
- What geographical regions have the strongest presence?
- What is the current trend in TWS group awareness?
- What factors and events correspond with changes in awaareness?
- How do consumers engage with TWS?
- Which platforms are most commonly used by TWS's audience?
- Which marketing channels were utilized during comeback campaigns?
- Were there any consumer engagement trends throughout the comeback campaign period?
- Which audience segments were most engaged in the comeback?
- How does TWS compare with competitors?
